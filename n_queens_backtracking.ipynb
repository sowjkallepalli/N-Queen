Solving n - queens problem using backtracking

```python
# Change n value accordingly
n = 22
```

```python
from typing import List

### The code stops backtracking after 4 solutions are found.

def nQueen(n : int) -> List[List[str]]:
    visited_col = set()
    posDiag = set()
    negDiag = set()

    #to store the result
    result = []
    chess_board = [["."] * n for _ in range(n)]

    #  row by row until we reach n                         
    def backtrack(row):
        if row == n :
            copy = []
            for row_num in range(n):
                for col_num in range(n):
                    if chess_board[row_num][col_num] == 'Q':
                        copy.append((row_num,col_num))
            result.append(copy)
            return
      
        for col in range(n):
          if col in visited_col or (row+col) in posDiag or (row-col) in negDiag:
            continue

          visited_col.add(col)
          posDiag.add(row+col)
          negDiag.add(row-col)
          chess_board[row][col] = "Q"

          # This stopper check is to stop further backtracking after found 4 solutions.
          # To get all the results, need to remove this if condition.
          if len(result) == 4:
            return
          backtrack(row +1)

          visited_col.remove(col)
          posDiag.remove(row+col)
          negDiag.remove(row-col)
          chess_board[row][col] = "."
    backtrack(0)
    return result
```

```python
result = nQueen(n)
```

```python
def print_result(result):
   for solution in result:
      print(solution)
```
